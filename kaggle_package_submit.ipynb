# Package & submit

Run this in a **Kaggle notebook** once you have a trained `lora_adapter/` — either downloaded from Modal and uploaded as a Kaggle **dataset**, or produced in this notebook. It finds the adapter, builds `submission.zip` (files at the zip root, r<=32 checked), and tells you how to submit.

## 1. Package -> /kaggle/working/submission.zip

```python
import glob, os, subprocess, sys
# clone the packaging script if it isn't already here
if not os.path.exists('scripts/05_package_submission.py'):
    subprocess.run('git clone -b build/nemotron-pipeline https://github.com/SebAustin/NVIDIA-Nemotron-Model-Reasoning-Challenge repo', shell=True, check=True)
    os.chdir('repo')
# locate the adapter: a dataset mount, or a local ./lora_adapter
cands = glob.glob('/kaggle/input/**/adapter_config.json', recursive=True)
if os.path.exists('lora_adapter/adapter_config.json'):
    cands.append('lora_adapter/adapter_config.json')
assert cands, "No adapter found — upload lora_adapter as a dataset, or download it from Modal first."
adapter_dir = os.path.dirname(cands[0]); print('adapter:', adapter_dir)
subprocess.run([sys.executable, 'scripts/05_package_submission.py',
                '--adapter-dir', adapter_dir,
                '--output', '/kaggle/working/submission.zip'], check=True)
```

## 2. Submit
This competition scores the **notebook output** `submission.zip`:
1. **Save Version** (Commit) this notebook.
2. Open the version → **Output** → **Submit** `submission.zip` to the competition.
